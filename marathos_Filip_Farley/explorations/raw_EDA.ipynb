```python
VOLUME_PATH = "/Volumes/marathos/default/raw/data"

spark.sql(f"LIST '{VOLUME_PATH}'").display()
```

### Read CSV file and view columns

```python
# Read from the CSV file in the volume
marathos_df = spark.read.csv(f"{VOLUME_PATH}/TWO_CENTURIES_OF_UM_RACES.csv", header=True, inferSchema=True)
marathos_df.display()
```

### Number of rows

```python
total_rows = marathos_df.count()
print(f"Total rows: {total_rows}")
```

### Number of columns

```python
total_columns = len(marathos_df.columns)
print(f"Total columns: {total_columns}")
```

### Check schema    / some wrong...

```python
marathos_df.printSchema()
```

### Descriptive summary of numerical fields

```python
marathos_df.summary().display()
```

```python
marathos_df.describe().display()
```

### count nulls for each field

```python
from pyspark.sql.functions import col, sum

null_values = marathos_df.select([sum(col(c).isNull().cast("int").alias(c)) for c in marathos_df.columns])

display(null_values)
```

### Unique Events

```python
unique_events = marathos_df.select("Event name").distinct().count()

print(f"Unique events: {unique_events}")
```

### Age distribution among runners

```python
athlete_age_df = marathos_df.groupby("Athlete age category").count().distinct()

athlete_age_df.display()
```

### Plot age distribution

```python
athlete_age_df_sorted = athlete_age_df.orderBy("count")
athlete_age_df_sorted.plot(kind="bar", x="Athlete age category", y="count")
```

## Which countries are most represented?

```python
athlete_country_df = marathos_df.groupby("Athlete country").count().orderBy("count", ascending=False).limit(10)

athlete_country_df.plot(kind="bar", x="Athlete country", y="count")
```
