```sql
%sql
-- Find out number of athletes per country
USE CATALOG marathos;
USE SCHEMA platinum;

SELECT
    athlete_country,
    COUNT(DISTINCT athlete_id) AS number_of_athletes
FROM 
    marathos.platinum.mart_overview
GROUP BY 
    athlete_country
ORDER BY 
    number_of_athletes DESC;

-- Genie was correct 
```

```sql
%sql
-- Find out top 10 fastest times in 50 km races
SELECT
    athlete_id,
    event_name,
    athlete_performance_value,
    performance_unit,
    athlete_gender
FROM
    marathos.platinum.mart_distance_events
WHERE
    distance_category = '50 km'
    AND athlete_performance_value > 0
ORDER BY
    athlete_performance_value ASC
LIMIT 10;

-- Stämmer med Genie
```
